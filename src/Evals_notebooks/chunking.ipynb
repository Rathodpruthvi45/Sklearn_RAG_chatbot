```python
from langchain_core.documents import Document
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
from data_ingesations.load_docs import load_web_page, URLS
from data_ingesations.chunking import create_chunks
```

```python
embedding = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
vectostore = Chroma(
    collection_name="sklearn_docs",
    embedding_function=embedding,
    persist_directory="./chroma_db",
)

```

```python
documents = []
document_ids = []

for doc_num, url in enumerate(URLS):
    print(f"loading the document {doc_num+1}: {url}")

    text = load_web_page(url)
    chunks = create_chunks(text)

    print(f"chunks created : {len(chunks)}")

    for doc_id, chunk in enumerate(chunks):
        doc = Document(
            page_content=chunk,
            metadata={"source": url, "doc_num": doc_num, "chunk_num": doc_id},
        )
        documents.append(doc)
        document_ids.append(f"doc_{doc_num}_chunk_{doc_id}")
        
```

```python
vectostore.add_documents(
    documents=documents,
    ids=document_ids
)
```
